# 1 What is an API?
API stands for Application Programming Interface.

An API is a defined interface that allows two software applications to communicate with each other.

It specifies:
- What requests an application can make.
- What information it must provide.
- How the receiving application processes the request.
- What response it receives.

# 2 Client-server architecture

Before learning REST APIs, you need to understand client-server architecture.

It is a computing architecture in which two primary components communicate:
- Client: Requests information or a service.
- Server: Receives the request, processes it and returns a response.
Example
When you access a website:
- Your browser is the client.
- The website's backend is the server.
- The browser sends an HTTP request.
- The server returns an HTTP response.


Important terminology
- Request: Information sent from the client to the server.
- Response: Information returned by the server.
- Endpoint: A specific URL where an API resource or operation can be accessed.
- Payload: Data included in a request or response body.
- Protocol: Rules governing communication between systems.

# 3 What is REST?     

REST stands for Representational State Transfer.

It is an architectural style used to design network-based APIs.

A REST API generally exposes resources through URLs and uses HTTP methods to perform operations on those resources.

REST principles to understand
1. Client-server separation: Client and server have separate responsibilities.
2. Statelessness: Each request contains the information necessary for the server to process it. The server should not depend on conversational session state from previous requests.
3. Uniform interface: Resources are accessed through a consistent interface.
4. Cacheability: Responses can indicate whether they can be cached.
5. Layered system: A client does not need to know every internal layer between itself and the server.

# 3.1 What is HTTP?

HTTP stands for Hypertext Transfer Protocol.

It defines how clients and servers exchange requests and responses.

Notice that an HTTP exchange has two parts:
- Request: Method, URL, headers and optionally a body.
- Response: Status code, headers and optionally a body.

# 4 HTTP methods: GET, POST, PUT, PATCH and DELETE

```python
## GET — Retrieve data

import requests

url = "https://api.example.com/customers/101"

url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.get(url, timeout=10)
response.raise_for_status()

print(response.status_code)
print(response.json())

#The requests library sends an HTTP GET request.

#response.json() converts a JSON response into Python objects, typically dictionaries and lists.
```

```python
# 4.2 POST — Create data

import requests

url = "https://api.example.com/customers"

customer = {
    "name": "Sneha",
    "city": "Pune"
}

url = "https://jsonplaceholder.typicode.com/users"
response = requests.post(url, json=customer, timeout=10)
response.raise_for_status()

print(response.status_code)
print(response.json())
```

```python
## 4.3 PUT — Replace data

url = "https://api.example.com/customers/101"

updated_customer = {
    "name": "Rahul Sharma",
    "city": "Mumbai"
}

url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.put(url, json=updated_customer, timeout=10)

response.raise_for_status()

print(response.status_code)
print(response.json())
```

```python
## 4.4 PATCH — Partially update data

url = "https://api.example.com/customers/101"

update_data = {
    "city": "Bangalore"
}

url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.put(url, json=update_data, timeout=10)

response.raise_for_status()

print(response.status_code)
print(response.json())
```

```python
## 4.5 DELETE — Remove data
url = "https://api.example.com/customers/101"

url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.put(url, json=update_data, timeout=10)

response = requests.delete(url)

print(response.status_code)
```

The server may return:
- 200 OK with a response body.
- 202 Accepted if deletion is accepted but not completed.
- 204 No Content if deletion succeeds without a response body.

Important status codes
Code	Meaning	Example

```python
## Coding: Handle HTTP errors
import requests

url = "https://api.example.com/customers"

try:
    response = requests.get(url, timeout=10)

    response.raise_for_status()

    data = response.json()

    print("API request successful")
    print(data)

except requests.exceptions.HTTPError as error:
    print("HTTP error:", error)

except requests.exceptions.Timeout:
    print("API request timed out")

except requests.exceptions.ConnectionError:
    print("Unable to connect to API")

except requests.exceptions.RequestException as error:
    print("Request failed:", error)
```
